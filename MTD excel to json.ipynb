```python
import pandas as pd
import json
from pathlib import Path

# ===== CONFIG =====
EXCEL_PATH = r"F:\January'25\salesapp\glossery\testing\MTD.xlsx"
OUTPUT_JSON = r"F:\January'25\salesapp\glossery\testing\MTD.json"
SHEET_NAME = 0   # 0 = first sheet, or put sheet name string
# ==================

def excel_to_json():
    # Read Excel file
    df = pd.read_excel(
        EXCEL_PATH,
        sheet_name=SHEET_NAME,
        usecols="A:E",
        dtype=str  # read everything as string to avoid NaN issues
    )

    # Rename columns to clean keys
    df.columns = ["retailerId", "retailernum", "etop", "vas", "zl"]

    # Drop rows without retailerId
    df = df.dropna(subset=["retailerId"])

    # Strip whitespace
    df = df.applymap(lambda x: x.strip() if isinstance(x, str) else x)

    # Convert numeric fields if possible
    # Convert numeric fields ONLY
    # Convert numeric fields ONLY
    for col in ["etop", "vas", "zl"]:
        df[col] = pd.to_numeric(df[col], errors="coerce").fillna(0)

# Make sure retailer number is string
    df["retailernum"] = df["retailernum"].astype(str)


    # Convert to list of dicts
    data = df.to_dict(orient="records")

    # Write JSON
    Path(OUTPUT_JSON).parent.mkdir(parents=True, exist_ok=True)
    with open(OUTPUT_JSON, "w", encoding="utf-8") as f:
        json.dump(data, f, indent=4)

    print(f"✅ Successfully created JSON file: {OUTPUT_JSON}")
    print(f"📊 Records exported: {len(data)}")


if __name__ == "__main__":
    excel_to_json()
```
